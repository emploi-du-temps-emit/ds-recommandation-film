# Phase 1 : Importation des données

**Objectif :** Charger les fichiers CSV du dataset MovieLens et explorer la structure des données.

## Étapes :
1. Charger movies.csv
2. Charger ratings.csv  
3. Afficher les premières lignes
4. Vérifier les types et dimensions

```python
# Importation des bibliothèques
import pandas as pd
import numpy as np

# Configuration d'affichage
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 1000)
pd.set_option('display.max_rows', 20)

print('✅ Bibliothèques importées avec succès')
```

```python
# Chargement des fichiers CSV
movies = pd.read_csv('../data/ml-latest-small/movies.csv')
ratings = pd.read_csv('../data/ml-latest-small/ratings.csv')
tags = pd.read_csv('../data/ml-latest-small/tags.csv')
links = pd.read_csv('../data/ml-latest-small/links.csv')

print('✅ Fichiers chargés avec succès')
```

```python
# === MOVIES ===
print('='*60)
print("MOVIES.CSV")
print('='*60)
print(f"Dimensions : {movies.shape}")
print(f"Colonnes : {list(movies.columns)}\n")
print("Aperçu :")
display(movies.head(10))
print(f"\nTypes :\n{movies.dtypes}")
```

```python
# === RATINGS ===
print('='*60)
print("RATINGS.CSV")
print('='*60)
print(f"Dimensions : {ratings.shape}")
print(f"Colonnes : {list(ratings.columns)}\n")
print("Aperçu :")
display(ratings.head(10))
print(f"\nTypes :\n{ratings.dtypes}")
```

```python
# Statistiques rapides
print(f"🎬 Nombre de films : {movies['movieId'].nunique()}")
print(f"👥 Nombre d'utilisateurs : {ratings['userId'].nunique()}")
print(f"⭐ Nombre d'évaluations : {len(ratings)}")
print(f"📊 Note moyenne : {ratings['rating'].mean():.2f}/5")
print(f"📈 Note médiane : {ratings['rating'].median():.2f}/5")
```

```python
# Sauvegarde des données brutes pour les phases suivantes
movies.to_csv('../data/movies_clean.csv', index=False)
ratings.to_csv('../data/ratings_clean.csv', index=False)

print('✅ Données sauvegardées dans data/')
```
