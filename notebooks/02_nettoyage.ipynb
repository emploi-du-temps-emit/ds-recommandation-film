# Phase 2 : Nettoyage des données

**Objectif :** Nettoyer et préparer les données pour l'analyse.

## Étapes :
1. Vérifier les valeurs manquantes
2. Supprimer les doublons
3. Vérifier les plages de valeurs
4. Fusionner les données movies + ratings

```python
import pandas as pd
import numpy as np

# Chargement des données
movies = pd.read_csv('../data/movies_clean.csv')
ratings = pd.read_csv('../data/ratings_clean.csv')

print(f'Movies : {movies.shape}')
print(f'Ratings : {ratings.shape}')
```

```python
# 1. VALEURS MANQUANTES
print('=== Valeurs manquantes ===')
print('\nMovies :')
print(movies.isnull().sum())
print('\nRatings :')
print(ratings.isnull().sum())
```

```python
# 2. DOUBLONS
print('=== Doublons ===')
print(f'Doublons dans movies : {movies.duplicated().sum()}')
print(f'Doublons dans ratings : {ratings.duplicated().sum()}')

# Supprimer les doublons si nécessaire
movies = movies.drop_duplicates(subset=['movieId'])
ratings = ratings.drop_duplicates(subset=['userId', 'movieId'])

print(f'Après nettoyage - Movies : {len(movies)}, Ratings : {len(ratings)}')
```

```python
# 3. PLAGES DE VALEURS
print('=== Vérification des notes ===')
print(f'Min : {ratings["rating"].min()}')
print(f'Max : {ratings["rating"].max()}')
print(f'Valeurs uniques : {sorted(ratings["rating"].unique())}')

# Filtrer les notes invalides (si nécessaire)
ratings = ratings[(ratings['rating'] >= 0.5) & (ratings['rating'] <= 5.0)]
print(f'Après filtrage : {len(ratings)} évaluations')
```

```python
# 4. FUSION DES DONNÉES
data = pd.merge(ratings, movies, on='movieId', how='left')
print(f'DataFrame fusionné : {data.shape}')
print(f'Valeurs manquantes après fusion :\n{data.isnull().sum()}')

# Supprimer les lignes sans film
data = data.dropna(subset=['title'])
print(f'Après suppression des NaN : {len(data)}')
```

```python
# Aperçu final
display(data.head())
print(f'\nDimensions finales : {data.shape}')
print(f'Utilisateurs : {data["userId"].nunique()}')
print(f'Films : {data["movieId"].nunique()}')
```

```python
# Sauvegarde
data.to_csv('../data/ratings_with_movies.csv', index=False)
print('✅ Données nettoyées sauvegardées')
```
